```python
import torch

# 1. Define probabilities for each category (must sum to 1)
probs = torch.ones(100)

# 2. Draw 200k samples
# replacement=True allows categories to be drawn multiple times per draw

samples = torch.tensor([])
for i in range(200000):
    num_samples = 1
    samples = torch.cat((samples, torch.multinomial(probs, num_samples, replacement=True)), dim=0)

print(samples.shape, samples[:100])
num_samples = 200000
# 3. Analyze the results (count the occurrences of each category)
unique, counts = torch.unique(samples, return_counts=True)
frequencies = counts / num_samples

for i in range(len(unique)):
    print(f"Category {unique[i].item()}: {frequencies[i].item():.4f}")
```

```python


# draw histogram
# print(f"avg = {num_steps * batch_size / Xtrain.shape[0]}")
data = frequencies
plt.figure(figsize=(8, 5))
plt.hist(data, bins=500, density=True, alpha=0.6, color='g', edgecolor='black', label='Sample Histogram')

plt.title('Empirical PDF via Normalized Histogram')
plt.xlabel('Values')
plt.ylabel('Probability Density')
plt.legend()
plt.show()
```
